# SmartBusiness – Cola de pedidos

**Estructura de datos: Cola FIFO (First In, First Out)**

Este notebook contiene un programa para administrar pedidos de clientes. Los pedidos se atienden en el mismo orden en que llegan: el primero que entra es el primero que sale.

La explicación de cada sección aparece antes de su código. Ejecuta las celdas en orden, desde la primera hasta la última.


## 1. Crear la cola

Creamos una lista vacía llamada `cola_pedidos`. En este programa la utilizamos como una cola: los nuevos pedidos se agregan al final y se procesan desde el principio.

```python
cola_pedidos = []
```

## 2. Función para agregar pedidos

La función `agregar_pedido()` solicita el nombre del cliente, el producto y la cantidad. Después guarda esos datos en un diccionario y agrega el diccionario al final de la cola mediante `append()`.

```python
def agregar_pedido():
    cliente = input("Nombre del cliente: ")
    producto = input("Producto: ")
    cantidad = int(input("Cantidad: "))

    pedido = {
        "cliente": cliente,
        "producto": producto,
        "cantidad": cantidad
    }

    cola_pedidos.append(pedido)

    print("Pedido agregado correctamente.")
```

## 3. Función para procesar un pedido

Primero verificamos si la cola está vacía. Si no hay pedidos, mostramos un mensaje y terminamos la función con `return`.

Si hay pedidos, usamos `pop(0)` para obtener y eliminar el elemento de la posición 0. Así procesamos siempre el pedido más antiguo y se cumple el principio FIFO.

```python
def procesar_pedido():
    if not cola_pedidos:
        print("No hay pedidos pendientes.")
        return

    pedido = cola_pedidos.pop(0)

    print("\nPedido procesado:")
    print("Cliente:", pedido["cliente"])
    print("Producto:", pedido["producto"])
    print("Cantidad:", pedido["cantidad"])
```

## 4. Función para ver el siguiente pedido

Esta función permite consultar el primer pedido pendiente sin eliminarlo. Para hacerlo usamos `cola_pedidos[0]`. A diferencia de `pop(0)`, esta instrucción solamente consulta el elemento.

```python
def ver_siguiente():
    if not cola_pedidos:
        print("No hay pedidos pendientes.")
        return

    pedido = cola_pedidos[0]

    print("\nSiguiente pedido:")
    print("Cliente:", pedido["cliente"])
    print("Producto:", pedido["producto"])
    print("Cantidad:", pedido["cantidad"])
```

## 5. Función para mostrar toda la cola

La función `mostrar_cola()` muestra todos los pedidos que siguen pendientes. `for` recorre la lista y `enumerate(cola_pedidos, 1)` permite mostrar la posición de cada pedido comenzando desde 1.

```python
def mostrar_cola():
    if not cola_pedidos:
        print("La cola está vacía.")
        return

    print("\n--- PEDIDOS PENDIENTES ---")

    for posicion, pedido in enumerate(cola_pedidos, 1):
        print(
            posicion,
            "- Cliente:", pedido["cliente"],
            "| Producto:", pedido["producto"],
            "| Cantidad:", pedido["cantidad"]
        )
```

## 6. Menú principal

La función `menu()` controla el programa. El ciclo `while True` mantiene el menú en pantalla hasta que el usuario elija salir. `input()` recibe la opción y las condiciones `if`, `elif` y `else` llaman a la función correspondiente o muestran un mensaje si la opción no es válida.

```python
def menu():
    while True:

        print("\n==============================")
        print("       SMARTBUSINESS")
        print("==============================")
        print("1. Agregar pedido")
        print("2. Procesar pedido")
        print("3. Ver siguiente pedido")
        print("4. Mostrar cola")
        print("5. Salir")
        print("==============================")

        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            agregar_pedido()

        elif opcion == "2":
            procesar_pedido()

        elif opcion == "3":
            ver_siguiente()

        elif opcion == "4":
            mostrar_cola()

        elif opcion == "5":
            print("Programa finalizado.")
            break

        else:
            print("Opción inválida.")
```

## 7. Ejecutar el programa

Hasta ahora definimos las funciones, pero todavía no iniciamos el menú. Al ejecutar `menu()` comienza la interacción con el usuario.


```python
menu()
```
