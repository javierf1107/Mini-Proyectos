# Sistema de Gestión de Pedidos

Este programa permite gestionar pedidos utilizando programación orientada a objetos,
herencia y una pila (stack).

El sistema permite:
- Agregar pedidos.
- Atender pedidos.
- Mostrar los pedidos.
- Verificar si la pila está vacía.
- Salir del programa.

## 1. Clase Padre: Pedido

La clase `Pedido` representa la información básica que tienen todos los pedidos.

Recibe dos datos:
- `numero`: número del pedido.
- `cliente`: nombre del cliente.

El método `mostrar_pedido()` permite mostrar estos datos en pantalla.

```python
class Pedido:

    def __init__(self, numero, cliente):
        self.numero = numero
        self.cliente = cliente

    def mostrar_pedido(self):
        print("Numero de pedido:", self.numero)
        print("Cliente:", self.cliente)
```

## 2. Clase hija: PedidoComida

`PedidoComida` hereda de la clase `Pedido`.

Además de los datos que ya tiene la clase padre, agrega el atributo `comida`.

Se utiliza `super()` para reutilizar el constructor y el método de la clase padre.

```python
class PedidoComida(Pedido):

    def __init__(self, numero, cliente, comida):
        super().__init__(numero, cliente)
        self.comida = comida

    def mostrar_pedido(self):
        super().mostrar_pedido()
        print("Tipo: Comida")
        print("Comida:", self.comida)
```

## 3. Clase hija: PedidoProducto

`PedidoProducto` también hereda de `Pedido`.

Esta clase agrega el atributo `producto` para almacenar el nombre del producto solicitado.

Al igual que `PedidoComida`, sobrescribe el método `mostrar_pedido()`.

```python
class PedidoProducto(Pedido):

    def __init__(self, numero, cliente, producto):
        super().__init__(numero, cliente)
        self.producto = producto

    def mostrar_pedido(self):
        super().mostrar_pedido()
        print("Tipo: Producto")
        print("Producto:", self.producto)
```

## 4. Creación de la pila

Se utiliza una lista de Python llamada `pila` para almacenar los pedidos.

Una pila funciona con el principio **LIFO (Last In, First Out)**,
lo que significa que el último pedido agregado será el primero en ser atendido.

Para agregar se utiliza `append()` y para retirar el último elemento se utiliza `pop()`.

```python
pila = []
```

## 5. Menú principal

Se utiliza un ciclo `while True` para mantener el programa funcionando
hasta que el usuario seleccione la opción de salir.

El menú tiene cinco opciones:
1. Llenar pila.
2. Atender pedido.
3. Mostrar pila.
4. Verificar si la pila está vacía.
5. Salir.

```python
while True:

    print("\n===== SISTEMA DE GESTION DE PEDIDOS =====")
    print("1. Llenar pila")
    print("2. Atender pedido")
    print("3. Mostrar pila")
    print("4. Verificar si la pila esta vacia")
    print("5. Salir")

    opcion = input("Seleccione una opcion: ")
```

## 6. Validación de la opción

El programa verifica que el usuario solamente pueda seleccionar
una opción válida del 1 al 5.

Si escribe otra opción, se muestra un mensaje de error y se vuelve
a mostrar el menú.

```python
if opcion not in ["1", "2", "3", "4", "5"]:
    print("Error: seleccione una opcion del 1 al 5.")
    continue
```

## 7. Llenar la pila

La opción 1 permite registrar nuevos pedidos.

Antes de guardar el pedido se realizan validaciones:
- El número no puede estar vacío.
- El número debe contener solamente números.
- El nombre del cliente no puede estar vacío.
- El nombre del cliente debe contener solamente letras.
- El usuario debe seleccionar si es un pedido de comida o de producto.

```python
if opcion == "1":

    while True:

        print("\n===== AGREGAR PEDIDO =====")

        while True:

            numero = input("Ingrese el numero del pedido: ")

            if numero == "":
                print("Error: el numero no puede estar vacio.")

            elif not numero.isdigit():
                print("Error: el numero debe contener solamente numeros.")

            else:
                break

        while True:

            cliente = input("Ingrese el nombre del cliente: ")

            if cliente == "":
                print("Error: el cliente no puede estar vacio.")

            elif not cliente.replace(" ", "").isalpha():
                print("Error: el nombre debe contener solamente letras.")

            else:
                break

        while True:

            print("\nTipo de pedido:")
            print("1. Comida")
            print("2. Producto")

            tipo = input("Seleccione el tipo: ")

            if tipo == "1" or tipo == "2":
                break

            else:
                print("Error: seleccione 1 o 2.")


        if tipo == "1":

            while True:

                comida = input("Ingrese la comida: ")

                if comida == "":
                    print("Error: la comida no puede estar vacia.")

                else:
                    break

            pedido = PedidoComida(numero, cliente, comida)

        else:

            while True:

                producto = input("Ingrese el producto: ")

                if producto == "":
                    print("Error: el producto no puede estar vacio.")

                else:
                    break

            pedido = PedidoProducto(numero, cliente, producto)

        pila.append(pedido)

        print("\nPedido agregado correctamente!")
        print("Cantidad de pedidos en la pila:", len(pila))

        while True:

            continuar = input(
                "Desea agregar otro pedido? (s/n): "
            ).lower()

            if continuar == "s" or continuar == "n":
                break

            else:
                print("Error: escriba solamente s o n.")


        if continuar == "n":
            break
```

## 8. Atender pedido

La opción 2 permite atender un pedido.

Primero se verifica si la pila está vacía.

Si contiene pedidos, se utiliza `pop()` para sacar el último pedido agregado.
Esto demuestra el funcionamiento de una pila mediante el principio LIFO.

```python
if len(pila) == 0:

    print("\nLa pila esta vacia.")
    print("No hay pedidos para atender.")

else:

    pedido = pila.pop()

    print("\n===== PEDIDO ATENDIDO =====")

    pedido.mostrar_pedido()

    print("\nPedidos restantes:", len(pila))
```

## 9. Mostrar pedidos

La opción 3 permite visualizar todos los pedidos registrados.

Se utiliza `reversed(pila)` para mostrar primero el último pedido
agregado y después los anteriores, siguiendo el comportamiento de una pila.

```python
for pedido in reversed(pila):

    pedido.mostrar_pedido()

    print("------------------------")
```

## 10. Verificar si la pila está vacía

La opción 4 permite comprobar si existen pedidos dentro de la pila.

Se utiliza `len(pila)` para conocer la cantidad de elementos almacenados.

```python
if len(pila) == 0:

    print("\nLa pila esta VACIA.")

else:

    print("\nLa pila NO esta vacia.")
    print("Cantidad de pedidos:", len(pila))
```

## 11. Salir del programa

La opción 5 finaliza la ejecución del programa.

Se utiliza `break` para detener el ciclo `while True`.

```python
print("\nPrograma finalizado.")
break
```
